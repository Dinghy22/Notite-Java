# `printf()` - Formatted Output

**`printf()`** is used to print formatted text.

```java
System.out.printf("Hello %s!\n", name);
```

<span style="background-color:#fff3cd; color:#5f3b00; padding:2px 6px; border-radius:4px;">
Main idea: printf lets you control how values appear in the output.
</span>

---

## Basic Pattern

```java
System.out.printf("format text", value1, value2);
```

The format placeholder looks like this:

```txt
%[flags][width][.precision][specifier]
```

Example:

```java
System.out.printf("You are %.2f cm tall", height);
```

Output:

```txt
You are 120.30 cm tall
```

---

## Common Specifiers

| Specifier | Data Type | Example |
|---|---|---|
| `%s` | String | `"Ironman"` |
| `%c` | char | `'I'` |
| `%d` | int | `25` |
| `%f` | double / float | `120.3` |
| `%b` | boolean | `true` |

---

<details>
<summary><strong>Precision</strong></summary>

Precision controls how many decimal digits are shown.

```java
double price = 19.999;

System.out.printf("%.2f", price);
```

Output:

```txt
20.00
```

`printf()` rounds the number automatically.

</details>

<details>
<summary><strong>Useful Flags</strong></summary>

| Flag | Meaning | Example |
|---|---|---|
| `+` | Shows plus sign for positive numbers | `+120.30` |
| `,` | Adds thousands separators | `1,000,000` |
| `(` | Shows negative numbers in parentheses | `(120.30)` |
| space | Adds a space before positive numbers | ` 120.30` |

Example:

```java
System.out.printf("%,.2f", 1000000.5);
```

Output:

```txt
1,000,000.50
```

</details>

<details>
<summary><strong>Width and Alignment</strong></summary>

Width reserves space for the output.

```java
System.out.printf("%10s", "Java");
```

Output:

```txt
      Java
```

Use `-` for left alignment:

```java
System.out.printf("%-10s", "Java");
```

Output:

```txt
Java      
```

</details>

---

## Quick Examples

```java
String name = "Ironman";
char firstLetter = 'I';
double height = -120.3;
boolean isEmployed = true;

System.out.printf("Hello %s!\n", name);
System.out.printf("First letter: %c\n", firstLetter);
System.out.printf("Height: %.2f cm\n", height);
System.out.printf("Employed: %b\n", isEmployed);
```

---

## Remember

- Use `%s` for text.
- Use `%d` for whole numbers.
- Use `%f` for decimal numbers.
- Use `%.2f` for 2 decimal places.
- Use `\n` to move to a new line.

````java
String name = "Ironman";
char firstLetter = 'I';
double height = -120.3;
boolean isEmployed = true;
System.out.printf("Hello %s!\n",name);
System.out.printf("Your name s first letter is %c\n",firstLetter);
System.out.printf("You are % .2fcm tall\n",height);
````

````java
// Nested if statements 

boolean isStudent = true;
boolean isSenior = false;
double pret = 9.99;
if(isStudent){
    if(isSenior){
        System.out.println("Ai 10% reducere la bilet");
        System.out.println("Ai 20% reducere la bilet");
        pret *= 0.7;
    }
    else{
        System.out.println("Ai 10% reducere la bilet");
        pret *= 0.9;
    }
    
}
else{
    if(isSenior){
        System.out.println("Ai 20% reducere la bilet");
        pret *= 0.8;
    }
    
}

System.out.printf("Pretul biletului este %.2f",pret);
````

# `String` Methods

**`String` methods** are built-in actions that let you inspect, compare, and transform text.

```java
String nume = "Dinca Teodor";
```

<span style="background-color:#e8f4ff; color:#12395c; padding:2px 6px; border-radius:4px;">
Main idea: Strings are indexed starting from 0.
</span>

---

## Basic Methods

| Method | What It Does | Example | Result |
|---|---|---|---|
| `.length()` | Counts characters | `nume.length()` | `12` |
| `.charAt(index)` | Gets character at index | `nume.charAt(10)` | `'o'` |
| `.indexOf(value)` | Finds first position | `nume.indexOf("o")` | `8` |
| `.lastIndexOf(value)` | Finds last position | `nume.lastIndexOf("o")` | `10` |
| `.toUpperCase()` | Converts to uppercase | `nume.toUpperCase()` | `"DINCA TEODOR"` |
| `.toLowerCase()` | Converts to lowercase | `nume.toLowerCase()` | `"dinca teodor"` |
| `.trim()` | Removes spaces at start/end | `nume.trim()` | `"Dinca Teodor"` |
| `.replace(old, new)` | Replaces matching text | `nume.replace("o", "a")` | `"Dinca Teadar"` |
| `.isEmpty()` | Checks if string is empty | `nume.isEmpty()` | `false` |
| `.contains(value)` | Checks if text exists inside | `nume.contains(" ")` | `true` |
| `.equals(value)` | Compares exactly | `nume.equals("Dinca Teodor")` | `true` |
| `.equalsIgnoreCase(value)` | Compares ignoring case | `nume.equalsIgnoreCase("dinca teodor")` | `true` |

## Remember

- Use `.length()` to count characters.
- Use `.charAt(index)` to get one character.
- Use `.indexOf()` and `.lastIndexOf()` to find positions.
- Use `.contains()` when you only need `true` or `false`.
- Use `.equals()` to compare strings, not `==`.
- Use `.substring()` to extract part of a string.

````java
// String methods

// string length
String nume = "Dinca Teodor";
System.out.println(nume.length());
// charAt
System.out.println(nume.charAt(10));
// first index of a character
System.out.println(nume.indexOf("o"));
// last index of a character
System.out.println(nume.lastIndexOf("o"));
// thransform a string to uppercase letters/ lowercase letters
System.out.println(nume.toUpperCase());
// trim/elimiate white spaces before and after a string
System.out.println(nume.trim());
// replace one character with another(all of them)
System.out.println(nume.replace("o","a"));
// check if a string is empty(boolean)
System.out.println(nume.isEmpty());
// check if a string contains any characters(boolean)
System.out.println(nume.contains(" "));
// check if two string are equal(case sensitive)
System.out.println(name.equals("Dinca Teodor"));
System.out.println(name.equalsIgnoreCase("dinca teodor"));
````

# `substring()` - String Slicing

**String slicing** means extracting part of a string.

```java
String email = "teodordinca@gmail.com";
```

<span style="background-color:#fff3cd; color:#5f3b00; padding:2px 6px; border-radius:4px;">
In Java, use `.substring()` for slicing. Java does not support negative indexes.
</span>

---

## Basic Syntax

```java
string.substring(start, end)
```

- `start` = index where slicing begins
- `end` = index where slicing stops
- The `end` index is **not included**

```java
String username = email.substring(0, email.indexOf("@"));
String domain = email.substring(email.indexOf("@") + 1);
```

Output:

```txt
teodordinca
gmail.com
```

---

# Exception Handling in Java

**Exception handling** inseamna tratarea erorilor care apar in timpul rularii programului, fara ca programul sa se opreasca brusc.

```java
try{
    // cod care poate produce o eroare
}
catch (Exception ex){
    // cod care se executa daca apare eroarea
}
```

<span style="background-color:#fff3cd; color:#5f3b00; padding:2px 6px; border-radius:4px;">
Ideea principala: o exceptie este un obiect care descrie o problema aparuta in timpul executiei.
</span>

---

## Cum functioneaza `try` si `catch`

| Parte | Ce face | Exemplu |
|---|---|---|
| `try` | contine codul riscant | accesarea unui index dintr-un vector |
| `catch` | prinde si trateaza exceptia | `catch(ArrayIndexOutOfBoundsException ex)` |
| `ex` | obiectul exceptiei | poate contine mesajul si cauza erorii |

In exemplul de mai jos, vectorul are doar 2 pozitii: `0` si `1`. Daca incercam sa folosim indexul `2`, `3`, `4` etc., Java arunca o exceptie.

```java
int[] vector = new int[2];

try{
    for(int i = 0; i < 10; i++){
        vector[i] = 3;
    }
}
catch(ArrayIndexOutOfBoundsException ex){
    System.out.println("s-a depasit indexul vectorului");
}
```

Output:

```txt
s-a depasit indexul vectorului
```

---

## 5 exceptii frecvente in Java

| Exceptie | Cand apare de obicei | Exemplu scurt |
|---|---|---|
| `NullPointerException` | folosesti un obiect care are valoarea `null` | `nume.length()` cand `nume == null` |
| `ArrayIndexOutOfBoundsException` | accesezi un index invalid intr-un vector | `vector[5]` cand vectorul are doar 2 elemente |
| `ArithmeticException` | operatie matematica imposibila | impartire la `0` cu numere intregi |
| `NumberFormatException` | transformi text invalid in numar | `Integer.parseInt("abc")` |
| `FileNotFoundException` | programul cauta un fisier care nu exista | citirea unui fisier cu nume gresit |

---

<details>
<summary><strong>`finally`, `throw` si `throws`</strong></summary>

| Cuvant cheie | Ce face | Exemplu |
|---|---|---|
| `finally` | ruleaza indiferent daca apare sau nu exceptia | inchiderea unui fisier |
| `throw` | arunca manual o exceptie | `throw new IllegalArgumentException();` |
| `throws` | anunta ca metoda poate arunca o exceptie | `void citire() throws IOException` |

Exemplu cu `finally`:

```java
try{
    System.out.println("incerc sa rulez codul");
}
catch(Exception ex){
    System.out.println("a aparut o eroare");
}
finally{
    System.out.println("aceasta parte ruleaza mereu");
}
```

</details>

<details>
<summary><strong>Checked vs Unchecked Exceptions</strong></summary>

| Tip | Ce inseamna | Exemple |
|---|---|---|
| Checked exception | Java te obliga sa o tratezi sau sa o declari cu `throws` | `FileNotFoundException`, `IOException` |
| Unchecked exception | apare de obicei din greseli de logica si nu esti obligat sa o prinzi | `NullPointerException`, `ArithmeticException` |

</details>

---

## Cand nu stiu ce exceptie poate aparea?

Daca nu stii exact ce exceptie poate produce codul din `try`, poti folosi temporar `catch(Exception ex)` ca sa descoperi tipul problemei.

```java
try{
    // cod care poate produce o eroare
}
catch(Exception ex){
    System.out.println("Tip eroare: " + ex.getClass().getSimpleName());
    System.out.println("Mesaj: " + ex.getMessage());

    ex.printStackTrace();
}
```

| Linie | Ce iti arata |
|---|---|
| `ex.getClass().getSimpleName()` | numele exact al exceptiei |
| `ex.getMessage()` | mesajul scurt al erorii |
| `ex.printStackTrace()` | locul exact din cod unde s-a produs eroarea |

Dupa ce afli tipul exceptiei, este mai bine sa prinzi exceptia specifica:

```java
try{
    int[] vector = new int[2];
    System.out.println(vector[5]);
}
catch(ArrayIndexOutOfBoundsException ex){
    System.out.println("Index invalid in vector.");
}
```

Poti avea mai multe blocuri `catch`. Exceptiile specifice se pun primele, iar `Exception` se pune la final ca varianta generala.

```java
try{
    // cod riscant
}
catch(NumberFormatException ex){
    System.out.println("Textul nu poate fi transformat in numar.");
}
catch(ArrayIndexOutOfBoundsException ex){
    System.out.println("Index invalid.");
}
catch(Exception ex){
    System.out.println("Alta eroare: " + ex.getMessage());
}
```

<span style="background-color:#e8f4ff; color:#12395c; padding:2px 6px; border-radius:4px;">
Regula practica: foloseste `Exception` pentru investigare sau fallback, apoi treci la exceptii specifice cand intelegi eroarea.
</span>

---

## Remember

- `try` contine codul care poate produce o problema.
- `catch` prinde exceptia si iti permite sa controlezi reactia programului.
- Exceptia este un obiect, de aceea poate fi salvata intr-o variabila ca `ex`.
- Pune mai intai `catch` pentru exceptii specifice, apoi pentru exceptii generale.
- Exception handling nu repara automat codul, ci previne oprirea brusca a programului.

````java
// Exception handlling

int[] vector = new int[2];
try{
    for(int i=0; i<10; i++)
        vector[i]= 3;
}
// exceptia este un obiect!!!
catch (ArrayIndexOutOfBoundsException ex){
    System.out.println("s a depasit index ul");
}
````

# Introducere in OOP Java

**OOP** = **Object Oriented Programming**. In Java, folosim OOP ca sa grupam datele si comportamentul intr-o singura structura: **clasa**.

```java
class Elev{
    String nume;
    int varsta;

    void vorbeste(){
        System.out.println(nume + " vorbeste");
    }
}
```

<span style="background-color:#e8f4ff; color:#12395c; padding:2px 6px; border-radius:4px;">
Ideea principala: clasa este tiparul, iar obiectul este o instanta concreta creata dupa acel tipar.
</span>

---

## Principiile OOP

| Principiu | Explicatie scurta | Exemplu simplu |
|---|---|---|
| Incapsulare | Tin datele si metodele impreuna si controlez accesul la date | campuri `private` + metode `get` / `set` |
| Mostenire | O clasa poate prelua comportament de la alta clasa | `class Elev extends Persoana` |
| Polimorfism | Acelasi nume de metoda poate avea comportament diferit in clase diferite | `vorbeste()` la obiecte diferite |
| Abstractizare | Ascund detaliile complicate si folosesc doar ce este necesar | folosesc `getNume()` fara sa stiu logica interna |

---

## Clasa si Obiect

| Termen | Ce inseamna | In cod |
|---|---|---|
| Clasa | tiparul general al unei categorii de date | `class Elev { ... }` |
| Atribut / camp | caracteristica unui obiect | `String nume;` |
| Metoda | actiune/functie a obiectului | `void vorbeste(){ ... }` |
| Obiect | instanta concreta a clasei | `Elev elev1 = new Elev();` |

---

<details>
<summary><strong>De retinut</strong></summary>

- Clasa descrie ce poate avea si ce poate face un obiect.
- Obiectul este creat cu `new`.
- `this` inseamna "obiectul curent".
- OOP ajuta codul sa fie mai organizat, reutilizabil si mai usor de modificat.

</details>

````java
// In Java, clasele sunt initializate avand denumiri care incep cu litera mare
class Elev{
    // Caracteristici
    String nume;
    int varsta;
    // Metode
    void vorbeste(){
        System.out.printf("Elevul %s vorbeste ",nume);
    }
}
void main(){
    // Asa se creaza un obiect al unei clase - elev1:
    // Nume_Clasa nume_obiect = new Nume_Clasa();
    Elev elev1 = new Elev();

    // Modifc atributul(numele) obiectului elev1
    elev1.nume = "Dinca Teodor";
    // Acceses metoda obiectului elev1
    elev1.vorbeste();
}
main();
````

# Constructori in Java

**Constructorul** este blocul folosit pentru initializarea unui obiect atunci cand obiectul este creat cu `new`.

```java
Elev elev = new Elev("Dinca Teodor", 17);
```

<span style="background-color:#fff3cd; color:#5f3b00; padding:2px 6px; border-radius:4px;">
Ideea principala: un constructor pregateste obiectul cu valori initiale corecte.
</span>

---

## Forma de baza

Constructorul are acelasi nume ca si clasa si nu are tip de returnare.

```java
class Elev{
    String nume;

    Elev(String nume){
        this.nume = nume;
    }
}
```

---

## Tipuri de constructori

| Tip | Ce inseamna | Exemplu |
|---|---|---|
| Constructor default | nu primeste argumente | `Elev(){ }` |
| Constructor cu parametri | primeste valori pentru obiect | `Elev(String nume){ ... }` |
| Constructor overload | aceeasi clasa are mai multi constructori diferiti | `Elev()` si `Elev(String nume)` |

---

<details>
<summary><strong>Regula importanta</strong></summary>

Daca scrii cel putin un constructor cu parametri, Java nu mai genereaza automat constructorul default.

```java
class Elev{
    Elev(String nume){
        this.nume = nume;
    }
}

// Elev elev = new Elev(); // eroare daca nu exista constructor fara argumente
```

</details>

---

## Remember

- Constructorul se ruleaza automat cand creezi obiectul.
- `this.camp` se refera la campul obiectului curent.
- Constructorii ajuta obiectele sa porneasca intr-o stare corecta.
- Poti avea mai multi constructori daca au liste diferite de parametri.

````java
class Elev{
    // Caracteristici
    String nume;
    int varsta;
    
    // Constructori
    
    // Constructor default(fara argumente)
    Elev(){

    }
    // Constructor care accepta un argument
    Elev(String nume){
        // this.nume acceseaza atributul actual al obiectului pentru claritate la atribuire
        this.nume = nume;
        // nume = nume nu va functiona
    }
    // Constructor care accepta doua argumente
    Elev(String nume, int varsta){
        this.nume = nume;
        this.varsta = varsta;
    }
    
    
    // Metode
    void vorbeste(){
        System.out.printf("Elevul %s vorbeste\n",nume);
    }
}
void main(){
    
    Elev elev1 = new Elev();
    Elev elev2 = new Elev("Lionel Messi");
    Elev elev3 = new Elev("Yamil Angura",15);


    elev1.nume = "Dinca Teodor";
    
    elev1.vorbeste();
    System.out.println(elev2.nume);
    System.out.println(elev3.varsta);
}
main();
````

# Constructii statice in Java

**`static`** inseamna ca un camp sau o metoda apartine clasei, nu unui obiect separat.

```java
Elev.numar
Elev.afisare();
```

<span style="background-color:#e8f4ff; color:#12395c; padding:2px 6px; border-radius:4px;">
Ideea principala: membrii `static` sunt comuni pentru toata clasa.
</span>

---

## Obiect vs Static

| Tip de membru | Apartine cui? | Cum il accesez? | Exemplu |
|---|---|---|---|
| Camp normal | fiecarui obiect | prin obiect | `elev1.nume` |
| Metoda normala | fiecarui obiect | prin obiect | `elev1.vorbeste()` |
| Camp static | clasei | prin clasa | `Elev.numar` |
| Metoda static | clasei | prin clasa | `Elev.afisare()` |

---

<details>
<summary><strong>Cand folosesc static?</strong></summary>

- Cand vreau o valoare comuna pentru toate obiectele clasei.
- Cand vreau o metoda utilitara care nu depinde de datele unui obiect.
- Pentru constante globale ale clasei, de obicei impreuna cu `final`: `static final`.

</details>

---

## Remember

- Acceseaza membrii static prin numele clasei.
- Membrii static nu au nevoie de un obiect creat cu `new`.
- Campurile normale sunt separate pentru fiecare obiect.

````java
class Elev{
    // Caracteristici
    String nume;
    int varsta;
    // Metode
    void vorbeste(){
        System.out.printf("Elevul %s vorbeste\n",nume);
    }
    // Cracteristici statice
    static int numar = 0;
    // Metode statice
    static void afisare(){
        System.out.println("Numar elevi: "+numar);
    }
    Elev(){
        numar++;
    }
}
void main(){
    
    Elev elev1 = new Elev();

    
    elev1.nume = "Dinca Teodor";
    
    elev1.vorbeste();
    System.out.println(Elev.numar);
    Elev.afisare();
}
main();
````

# Constante si Elemente de Siguranta in OOP

**Siguranta in OOP** inseamna sa controlezi cine poate citi sau modifica datele unui obiect.

```java
private final int anNastere;
```

<span style="background-color:#fff3cd; color:#5f3b00; padding:2px 6px; border-radius:4px;">
Ideea principala: `private` controleaza accesul, iar `final` controleaza modificarea.
</span>

---

## `private`

`private` este un **modificator de acces**. Un camp, constructor sau o metoda marcata cu `private` poate fi folosita direct doar in interiorul clasei in care a fost declarata.

| Unde folosesc `private` | Ce face | Exemplu |
|---|---|---|
| Camp / atribut | ascunde data in interiorul clasei | `private String nume;` |
| Metoda | ascunde o functie interna a clasei | `private boolean numeValid(String nume){ ... }` |
| Constructor | controleaza modul in care se creeaza obiectele | `private Persoana(){ }` |

Accesul controlat se face prin metode:

```java
String getNume(){
    return nume;
}

void setNume(String nume){
    this.nume = nume;
}
```

<details>
<summary><strong>De ce folosesc private?</strong></summary>

- Ca sa protejez datele obiectului de modificari gresite.
- Ca sa pot valida valorile inainte sa le salvez.
- Ca sa schimb logica interna a clasei fara sa stric codul care foloseste clasa.
- Acesta este principiul de **incapsulare**.

</details>

---

## `final`

`final` inseamna ca elementul respectiv nu mai poate fi schimbat dupa ce a fost stabilit. Efectul depinde de locul unde il folosesti.

| Unde folosesc `final` | Ce inseamna | Exemplu |
|---|---|---|
| Variabila locala | valoarea poate fi atribuita o singura data | `final int an = 2007;` |
| Camp al clasei | campul trebuie initializat la declarare sau in constructor | `private final int anNastere;` |
| Metoda | metoda nu poate fi suprascrisa prin override | `final void afisare(){ ... }` |
| Clasa | clasa nu poate fi mostenita | `final class Persoana { ... }` |

Pentru constante globale ale clasei se foloseste des combinatia `static final`:

```java
static final String SPECIE = "om";
static final int VARSTA_MINIMA = 0;
```

<details>
<summary><strong>Exemplu important cu final</strong></summary>

Un camp `final` poate fi setat in constructor, apoi ramane neschimbat pentru acel obiect.

```java
class Persoana{
    private final int anNastere;

    Persoana(int anNastere){
        this.anNastere = anNastere;
    }
}
```

Dupa constructor, `anNastere` nu mai poate primi alta valoare.

</details>

---

## Remember

- `private` raspunde la intrebarea: **cine are acces?**
- `final` raspunde la intrebarea: **se mai poate schimba?**
- Pentru date importante, foloseste `private` + metode `get` / `set`.
- Pentru valori care nu trebuie schimbate, foloseste `final`.
- Pentru constante ale clasei, foloseste `static final`.

````java
class Persoana{
    // Campurile private nu pot fi accesate direct din afara clasei.
    private String nume;
    private final int anNastere;
    private boolean rege = false;

    Persoana(String nume, int anNastere){
        this.nume = nume;
        this.anNastere = anNastere;
    }

    String getNume(){
        if(rege){
            return "regele " + nume;
        }
        return nume;
    }

    void setNume(String nume){
        if(nume == null || nume.isBlank()){
            System.out.println("Numele nu poate fi gol.");
            return;
        }

        this.nume = nume;
    }

    int getAnNastere(){
        return anNastere;
    }

    void investire(){
        rege = true;
    }
}

class Elev extends Persoana{
    private String scoala;

    Elev(String nume, int anNastere, String scoala){
        super(nume, anNastere);
        this.scoala = scoala;
    }

    String getScoala(){
        return scoala;
    }

    void setScoala(String scoala){
        this.scoala = scoala;
    }
}

void main(){
    Elev elev = new Elev("Dinca", 2007, "Colegiul National");

    // Acces corect la campuri private: prin metode get/set.
    elev.setNume("Dinca Teodor");
    elev.investire();

    System.out.println(elev.getNume());
    System.out.println(elev.getAnNastere());
    System.out.println(elev.getScoala());

    // Aceste linii NU ar functiona, deoarece campurile sunt private/final:
    // elev.nume = "Alt nume";
    // elev.anNastere = 2008;
}
main();
````

# Relatia `has-a` intre clase

**`has-a`** inseamna ca o clasa **are**, **contine** sau **foloseste** obiecte create din alta clasa.

```txt
Biblioteca are carti.
Masina are motor.
Elev are ghiozdan.
```

<span style="background-color:#e8f4ff; color:#12395c; padding:2px 6px; border-radius:4px;">
Ideea principala: `has-a` nu foloseste `extends`; foloseste un camp care are ca tip o alta clasa.
</span>

---

## Forma de baza

Daca o biblioteca are carti, atunci clasa `Biblioteca` poate avea un camp de tip `Carte`.

```java
class Carte{
    String titlu;
}

class Biblioteca{
    Carte carte;
}
```

Asta inseamna:

```txt
Biblioteca has-a Carte
Biblioteca are o Carte
```

---

## `has-a` vs `is-a`

| Intrebare | Daca raspunsul este DA | Relatie | Exemplu |
|---|---|---|---|
| X este un Y? | folosesc mostenire | `is-a` | `Elev` este o `Persoana` |
| X are un Y? | folosesc campuri/obiecte | `has-a` | `Biblioteca` are `Carte` |

Nu scriem `Carte extends Biblioteca`, pentru ca o carte **nu este** o biblioteca. O carte doar apartine sau este gestionata de biblioteca.

---

## Denumiri pe care le poti intalni

| Denumire | Explicatie simpla | Exemplu |
|---|---|---|
| Asociere | doua clase lucreaza impreuna | `Elev` foloseste o `Carte` |
| Agregare | o clasa are alte obiecte, dar acestea pot exista si separat | `Biblioteca` are `Carte` |
| Compozitie | o clasa contine parti care depind puternic de ea | `Casa` are `Camera` |

Pentru nivelul tau de acum, poti retine simplu: **`has-a` = o clasa are un camp care este obiect din alta clasa**.

---

<details>
<summary><strong>Cine are responsabilitatea?</strong></summary>

In exemplul cu biblioteca:

- `Carte` stie date despre carte: titlu, autor, disponibilitate.
- `Biblioteca` stie date despre biblioteca: nume, status, numar de carti.
- `Biblioteca` poate imprumuta o carte si isi poate actualiza numarul de carti disponibile.
- `Carte` nu trebuie sa aiba metode precum `deschideBiblioteca()`, pentru ca nu este responsabilitatea cartii.

</details>

---

## Remember

- Folosesc `has-a` cand un obiect are nevoie de alt obiect ca parte din structura lui.
- `has-a` nu este acelasi lucru cu o clasa abstracta.
- `has-a` nu inseamna mostenire.
- Pentru `has-a`, pun un obiect ca atribut/camp intr-o alta clasa.

````java
class CarteBiblioteca{
    private String titlu;
    private String autor;
    private boolean disponibila = true;

    CarteBiblioteca(String titlu, String autor){
        this.titlu = titlu;
        this.autor = autor;
    }

    String getTitlu(){
        return titlu;
    }

    boolean esteDisponibila(){
        return disponibila;
    }

    void imprumuta(){
        disponibila = false;
    }

    void returneaza(){
        disponibila = true;
    }

    void afiseazaCarte(){
        System.out.println(titlu + " - " + autor);
        System.out.println("Disponibila: " + disponibila);
    }
}

class BibliotecaScolara{
    private String nume;
    private CarteBiblioteca[] carti = new CarteBiblioteca[10];
    private int numarCarti = 0;
    private int cartiDisponibile = 0;
    private boolean deschisa = true;

    BibliotecaScolara(String nume){
        this.nume = nume;
    }

    void adaugaCarte(CarteBiblioteca carte){
        if(numarCarti < carti.length){
            carti[numarCarti] = carte;
            numarCarti++;

            if(carte.esteDisponibila()){
                cartiDisponibile++;
            }
        }
    }

    void imprumutaCarte(CarteBiblioteca carte){
        if(!deschisa){
            System.out.println("Biblioteca este inchisa.");
            return;
        }

        if(carte.esteDisponibila()){
            carte.imprumuta();
            cartiDisponibile--;
            System.out.println("Carte imprumutata: " + carte.getTitlu());
        }
        else{
            System.out.println("Cartea este deja imprumutata.");
        }
    }

    void returneazaCarte(CarteBiblioteca carte){
        if(!carte.esteDisponibila()){
            carte.returneaza();
            cartiDisponibile++;
            System.out.println("Carte returnata: " + carte.getTitlu());
        }
    }

    void afiseazaStatus(){
        System.out.println("Biblioteca: " + nume);
        System.out.println("Deschisa: " + deschisa);
        System.out.println("Carti totale: " + numarCarti);
        System.out.println("Carti disponibile: " + cartiDisponibile);
    }
}

void main(){
    CarteBiblioteca carte1 = new CarteBiblioteca("Amintiri din copilarie", "Ion Creanga");
    CarteBiblioteca carte2 = new CarteBiblioteca("Baltagul", "Mihail Sadoveanu");

    BibliotecaScolara biblioteca = new BibliotecaScolara("Biblioteca scolii");
    biblioteca.adaugaCarte(carte1);
    biblioteca.adaugaCarte(carte2);

    biblioteca.afiseazaStatus();
    biblioteca.imprumutaCarte(carte1);
    biblioteca.afiseazaStatus();

    carte1.afiseazaCarte();
}
main();
````

# Mostenire in Java (`is-a`)

**Mostenirea** inseamna ca o clasa copil preia campuri si metode de la o clasa parinte.

```txt
Elev este o Persoana.
Profesor este o Persoana.
```

<span style="background-color:#fff3cd; color:#5f3b00; padding:2px 6px; border-radius:4px;">
Ideea principala: folosesc mostenire cand pot spune corect "X este un Y".
</span>

---

## Forma de baza

In Java, mostenirea se face cu `extends`.

```java
class Persoana{
    String nume;
}

class Elev extends Persoana{
    String clasa;
}
```

Asta inseamna:

```txt
Elev is-a Persoana
Elev este o Persoana
```

---

## Termeni importanti

| Termen | Ce inseamna | Exemplu |
|---|---|---|
| Clasa parinte / superclass | clasa generala | `Persoana` |
| Clasa copil / subclass | clasa mai specifica | `Elev` |
| `extends` | cuvantul care creeaza mostenirea | `class Elev extends Persoana` |
| `super(...)` | apeleaza constructorul clasei parinte | `super(nume, anNastere);` |
| Override | copilul rescrie o metoda mostenita | `afiseazaRol()` |

---

## Ce mosteneste clasa copil?

| Element din clasa parinte | Ce se intampla in copil |
|---|---|
| Campuri accesibile | pot fi folosite de clasa copil |
| Metode accesibile | pot fi apelate sau suprascrise |
| Constructori | nu se mostenesc direct, dar se apeleaza cu `super(...)` |
| Campuri `private` | exista in obiect, dar nu pot fi accesate direct din copil |

Daca parintele are campuri `private`, copilul foloseste metode `get` / `set` sau constructorul parintelui.

---

<details>
<summary><strong>Cand folosesc mostenire?</strong></summary>

Foloseste mostenire cand ai o clasa generala si clase mai specifice.

| Corect | De ce |
|---|---|
| `Elev extends Persoana` | un elev este o persoana |
| `Profesor extends Persoana` | un profesor este o persoana |
| `Carte extends Biblioteca` | gresit: o carte nu este o biblioteca |

</details>

<details>
<summary><strong>Reguli importante</strong></summary>

- In Java, o clasa poate avea un singur parinte direct.
- Daca parintele are constructor cu parametri, copilul trebuie sa il apeleze cu `super(...)`.
- O metoda `final` nu poate fi suprascrisa.
- O clasa `final` nu poate fi mostenita.
- Nu folosi mostenire doar ca sa refolosesti cod; foloseste-o cand relatia `is-a` este corecta.

</details>

---

## Remember

- Mostenirea este relatia `is-a`.
- `extends` leaga clasa copil de clasa parinte.
- `super(...)` trimite valori constructorului parinte.
- `private` ramane protejat: copilul nu acceseaza direct campurile private ale parintelui.
- Pentru relatie de tip "are", folosesc `has-a`, nu mostenire.

````java
class PersoanaScoala{
    private String nume;
    private final int anNastere;

    PersoanaScoala(String nume, int anNastere){
        this.nume = nume;
        this.anNastere = anNastere;
    }

    String getNume(){
        return nume;
    }

    int getAnNastere(){
        return anNastere;
    }

    void afiseazaDate(){
        System.out.println("Nume: " + nume);
        System.out.println("An nastere: " + anNastere);
    }

    void afiseazaRol(){
        System.out.println("Persoana din scoala");
    }
}

class ElevScoala extends PersoanaScoala{
    private String clasa;

    ElevScoala(String nume, int anNastere, String clasa){
        super(nume, anNastere);
        this.clasa = clasa;
    }

    @Override
    void afiseazaRol(){
        System.out.println("Elev in clasa " + clasa);
    }
}

class ProfesorScoala extends PersoanaScoala{
    private String materie;

    ProfesorScoala(String nume, int anNastere, String materie){
        super(nume, anNastere);
        this.materie = materie;
    }

    @Override
    void afiseazaRol(){
        System.out.println("Profesor de " + materie);
    }
}

void main(){
    ElevScoala elev = new ElevScoala("Dinca Teodor", 2007, "11A");
    ProfesorScoala profesor = new ProfesorScoala("Popescu Ana", 1985, "Java");

    elev.afiseazaDate();
    elev.afiseazaRol();

    profesor.afiseazaDate();
    profesor.afiseazaRol();
}
main();
````
